````python
r = np.logspace(1, 5, 200)
fig, ax = plt.subplots(figsize=(5, 3.2))
ax.loglog(r, (r / 1e3) ** -0.5, label=r"toy $\propto (r/10^3 R_g)^{-1/2}$")
ax.set_xlabel(r"$r / R_g$")
ax.set_ylabel("toy scale")
ax.set_title("First-light figure (replace with your ZTF–GW plot)")
ax.legend()
fig.tight_layout()
plt.show()
print("First light — stack looks good.")
````
